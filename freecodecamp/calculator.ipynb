```python
import math
num1 = float(input("Enter num1 here:"))

num2 = float(input("enter num2 here:"))

op = input("Enter your desired operator (+,*,-,/,**,//,sin,cos,tan):")

if op == "+":
    print("Result:",round(num1+num2,2))

elif op == "-":
    print("Result:",round(num1-num2,2))

elif op == "*":
     print("Result:",round(num1*num2,2))

elif op == "/":
    if(num2 == 0):
        print("Invalid number try another one or just quit")
    else:
        print("Result:",round(num1/num2,2))

elif op == "**":
    print("Result:",num1**num2)

elif op == "//":
    print("Result:",num1//num2)

elif op == "sin":
    print("Result:",math.sin(num1))
    print("Result:",math.sin(num2))

elif op == "cos":
    print("Result:",math.cos(num1))
    print("Result:",math.cos(num2))     

elif op == "tan":
    print("Result:",math.tan(num1))
    print("Result:",math.tan(num2))
else:
    print("Invalid operator")
```
